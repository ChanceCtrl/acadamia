```python
print(' shape of x: ' + str(x.shape) + ' and y: ' + str(y.shape))
```

```python
# Initialize vectors here
m = len(x)
X = np.zeros((m,3))
Y = np.zeros((m,1))
theta = np.zeros((3,1))
```

```python
# you code starts here

# Make some matrixs
X[:, 0] = 1
X[:, 1] = x[:, 0]
X[:, 2] = x[:, 0]**2

# Copy y into Y
Y[:, 0] = y[:, 0]

# theta = (X^T X)^(-1) X^T Y
theta = np.linalg.inv(X.T @ X) @ X.T @ Y

print("theta =")
print(theta)

# Calculate predictions
y_hat = X @ theta

# Calculate min cost
J = (1/(2*m)) * np.sum((y_hat - Y)**2)

print("minimum cost =", J)

# Plot da regression
plt.figure()
x_fit = np.linspace(0, 12, 200).reshape(-1, 1)
X_fit = np.hstack((np.ones((len(x_fit), 1)),
                   x_fit,
                   x_fit**2))

y_fit = X_fit @ theta

# Plot the baseline fella
plt.plot(x_fit, y_fit, 'r-', linewidth=2)

# Do some stupid work to get the baseline y-value at the orignal Xs
X_stem = np.hstack((
    np.ones((len(x), 1)),
    x.reshape(-1, 1),
    x.reshape(-1, 1)**2
))
y_baseline = X_stem @ theta

# Plot the stem fellas
for xi, yi, bi in zip(x, y, y_baseline):
    plt.vlines(xi, bi, yi, linestyles="--", color="g")
    plt.plot(xi, yi, "x", color="C0")

plt.xticks(range(0, 14, 2))
plt.show()
```
